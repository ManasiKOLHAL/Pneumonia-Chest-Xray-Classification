## 6. Load Training & Validation Data:

This step loads images from folders and creates:
- Training dataset
- Validation dataset

````python
from tensorflow.keras.preprocessing.image import ImageDataGenerator

img_size = 224
batch_size = 32

datagen = ImageDataGenerator(
    rescale=1./255,
    validation_split=0.2,
    rotation_range=15,
    zoom_range=0.2,
    width_shift_range=0.1,
    height_shift_range=0.1,
    horizontal_flip=True,
    fill_mode='nearest'
)
# Load training dataset
train_dir = "Dataset/chest_xray/train"   

train_data = datagen.flow_from_directory(
    train_dir,
    target_size=(img_size,img_size),
    batch_size=batch_size,
    class_mode='binary',
    subset='training'
)
# Load validation dataset
val_data = datagen.flow_from_directory(
    train_dir,
    target_size=(img_size,img_size),
    batch_size=batch_size,
    class_mode='binary',
    subset='validation'
)
````

## 7. Build CNN Model:

CNN extracts image features such as patterns, textures and infected regions from X-rays.

````python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout

def build_custom_cnn(input_shape=(224,224,3)):
    model = Sequential([
        # Convolutional Block 1
        Conv2D(32, (3,3), activation='relu', input_shape=input_shape),
        MaxPooling2D(2,2),
        Dropout(0.25),

        # Convolutional Block 2
        Conv2D(64, (3,3), activation='relu'),
        MaxPooling2D(2,2),
        Dropout(0.25),

        # Convolutional Block 3
        Conv2D(128, (3,3), activation='relu'),
        MaxPooling2D(2,2),
        Dropout(0.3),

        # Fully Connected Layers
        Flatten(),
        Dense(256, activation='relu'),
        Dropout(0.5),
        Dense(1, activation='sigmoid')  # binary classification
    ])
    return model
````

## 8. Creating model:

This step loads pretrained models:
- ResNet50
- MobileNet
- VGG16

These models already know image features from ImageNet.

````python
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Dense, Dropout, GlobalAveragePooling2D
from tensorflow.keras.applications import ResNet50, MobileNetV2, VGG16


# ResNet50

def build_resnet50():
    base = ResNet50(
        weights='imagenet',
        include_top=False,
        input_shape=(224,224,3)
    )

    base.trainable = False

    x = base.output
    x = GlobalAveragePooling2D()(x)
    x = Dense(128, activation='relu')(x)
    x = Dropout(0.5)(x)
    output = Dense(1, activation='sigmoid')(x)

    model = Model(inputs=base.input, outputs=output)
    return model


# MobileNetV2

def build_mobilenet():
    base = MobileNetV2(
        weights='imagenet',
        include_top=False,
        input_shape=(224,224,3)
    )

    base.trainable = False

    x = base.output
    x = GlobalAveragePooling2D()(x)
    x = Dense(128, activation='relu')(x)
    x = Dropout(0.5)(x)
    output = Dense(1, activation='sigmoid')(x)

    model = Model(inputs=base.input, outputs=output)
    return model



# VGG16

def build_vgg16():
    base = VGG16(
        weights='imagenet',
        include_top=False,
        input_shape=(224,224,3)
    )

    base.trainable = False

    x = base.output
    x = GlobalAveragePooling2D()(x)
    x = Dense(128, activation='relu')(x)
    x = Dropout(0.5)(x)
    output = Dense(1, activation='sigmoid')(x)

    model = Model(inputs=base.input, outputs=output)
    return model
````

````python
custom_cnn = build_custom_cnn()
resnet_model = build_resnet50()
mobilenet_model = build_mobilenet()
vgg16_model = build_vgg16()
````

## 9. Compile Models:
This step prepares all models for training and comparison.

````python
models = [custom_cnn, resnet_model, mobilenet_model, vgg16_model]

for model in models:
    model.compile(
        optimizer='adam',
        loss='binary_crossentropy',
        metrics=['accuracy']
    )
# Compile Custom CNN
custom_cnn.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# Compile ResNet50
resnet_model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# Compile MobileNet
mobilenet_model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# Compile VGG16
vgg16_model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

print("All Models Compiled Successfully!")
````

## 17. Prediction Function:

Selected model predicts:

- Normal
or
- Pneumonia

````python
def predict_image(model,img_path):
    img = image.load_img(img_path,target_size=(224,224))
    img = image.img_to_array(img)/255
    img = np.expand_dims(img,axis=0)

    pred = model.predict(img)

    if pred > 0.5:
        return "Pneumonia",pred[0][0]
    else:
        return "Normal",pred[0][0]
````

This project explored four deep learning architectures—Custom CNN, ResNet50, MobileNetV2, and VGG16—for classifying chest X-ray images into Normal and Pneumonia categories.

Based on the saved training histories, the highest validation accuracy among the four models was achieved by MobileNetV2 (94.44%), followed by VGG16 (92.43%), Custom CNN (87.06%), and ResNet50 (76.32%).
